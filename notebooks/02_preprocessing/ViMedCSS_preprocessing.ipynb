## Baseline PhoWhisper trên validation/test/hard

### Chạy trên validation

```python
import os
import pandas as pd
import torch
from pathlib import Path
from datasets import load_dataset

PROJECT_ROOT = Path('/content/Clinical-Ambient-Documentation-Assistant')
AUDIO_EXTRACT_DIR = PROJECT_ROOT / "data" / "vimedcss_audio"
AUDIO_EXTRACT_DIR.mkdir(parents=True, exist_ok=True)
os.chdir(PROJECT_ROOT)

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# 1. Tải audio từ HuggingFace (nếu chưa đủ)
print("Kiểm tra dữ liệu âm thanh...")
ds = load_dataset("tensorxt/ViMedCSS", split="validation", trust_remote_code=True)
for i, item in enumerate(ds):
    audio_path = AUDIO_EXTRACT_DIR / os.path.basename(item['audio']['path'])
    if not audio_path.exists():
        with open(audio_path, "wb") as f:
            f.write(item['audio']['bytes'])
    if i > 200: break # Tăng số lượng lên để demo hoặc xóa break để tải hết

# 2. Chuẩn hóa manifest
print("Chuẩn hóa manifest...")
for split in ['validation', 'test', 'hard']:
    path = PROJECT_ROOT / f'experiments/asr/vimedcss/manifests/vimedcss_{split}_raw.jsonl'
    if path.exists():
        df = pd.read_json(path, lines=True)
        col = 'audio_path' if 'audio_path' in df.columns else ('audio' if 'audio' in df.columns else None)
        if col:
            df.rename(columns={col: 'clean_audio_path'}, inplace=True)
        df['clean_audio_path'] = df['clean_audio_path'].apply(lambda x: str(AUDIO_EXTRACT_DIR / os.path.basename(x)))
        df.to_json(path, orient='records', lines=True, force_ascii=False)

# 3. Chạy Evaluation lần lượt cho cả 3 file
%env HF_TOKEN=YOUR_HUGGINGFACE_TOKEN

splits = ["validation", "test", "hard"]
for split in splits:
    print(f"\n--- Đang chạy Evaluation trên tập: {split.upper()} ---")
    manifest_path = f"experiments/asr/vimedcss/manifests/vimedcss_{split}_raw.jsonl"
    output_path = f"experiments/asr/vimedcss/baseline/phowhisper_medium_{split}_predictions.jsonl"

    !python scripts/asr_eval/run_whisper_transformers.py \
      --manifest {manifest_path} \
      --model vinai/PhoWhisper-medium \
      --output {output_path} \
      --device {device}
```
